##3. Análisis Univariado (Una sola variable)
Aquí analizamos una variable aislada para ver su comportamiento.

Ejemplo: ¿Cómo se distribuyen las ventas totales? (Histograma)

Ejemplo: ¿Cuántos productos hay por categoría? (Gráfico de barras)

```python
# ANALISIS UNIVARIADO: VARIABLES NUMERICAS

print("--- ESTADISTICAS DE LA VARIABLE NUMERICA (Total_Venta) ---")

# Calculo de estadisticos centrales
promedio = df['Total_Venta'].mean()
mediana = df['Total_Venta'].median()
desviacion = df['Total_Venta'].std()
minimo = df['Total_Venta'].min()
maximo = df['Total_Venta'].max()

# Imprimir resultados en consola con formato de dos decimales
print(f"Promedio de venta: ${promedio:.2f}")
print(f"Mediana (valor central): ${mediana:.2f}")
print(f"Desviacion Estandar (variabilidad): {desviacion:.2f}")
print(f"Rango de precios: de ${minimo:.2f} a ${maximo:.2f}")

# Creacion del espacio para graficos (1 fila, 2 columnas)
plt.figure(figsize=(16, 6))

# Grafico 1: Histograma con curva de densidad
# Este grafico muestra donde se concentran la mayoria de las ventas
plt.subplot(1, 2, 1)
sns.histplot(df['Total_Venta'], kde=True, bins=30, color='#1f77b4')
plt.title('Histograma: Distribucion de Ventas')
plt.xlabel('Monto de Venta ($)')
plt.ylabel('Frecuencia (Cantidad de facturas)')

# Grafico 2: Diagrama de Caja (Boxplot)
# Este grafico es fundamental para detectar valores atipicos (outliers)
# Los puntos fuera de la "caja" son ventas inusualmente altas o bajas
plt.subplot(1, 2, 2)
sns.boxplot(x=df['Total_Venta'], color='#ff7f0e')
plt.title('Boxplot: Deteccion de Valores Atipicos')
plt.xlabel('Monto de Venta ($)')

# Ajustar el diseño y mostrar
plt.tight_layout()
plt.show()


# ANALISIS UNIVARIADO: VARIABLES CATEGORICAS

print("\n--- ESTADISTICAS DE LA VARIABLE CATEGORICA (Categoria) ---")

# Calculo de frecuencias absolutas (conteo)
conteo_categorias = df['Categoria'].value_counts()

# Calculo de frecuencias relativas (porcentaje)
porcentaje_categorias = df['Categoria'].value_counts(normalize=True) * 100

# Imprimir tabla de frecuencias
print("Conteo por Categoria:")
print(conteo_categorias)
print("\nPorcentaje de participacion por Categoria:")
print(porcentaje_categorias.round(2))

# Creacion del grafico de barras
plt.figure(figsize=(10, 6))

# Grafico 3: Grafico de Barras (Countplot)
# Muestra visualmente que categoria tiene mas transacciones
# Ordenamos las barras de mayor a menor para facilitar la lectura
order = df['Categoria'].value_counts().index
sns.countplot(x='Categoria', data=df, order=order, palette='viridis', hue='Categoria', legend=False)

plt.title('Frecuencia de Transacciones por Categoria')
plt.xlabel('Categoria del Producto')
plt.ylabel('Cantidad de Ventas')

# Ajustar y mostrar
plt.tight_layout()
plt.show()
```

###Histograma + Boxplot combinados:
En el análisis numérico, no basta con ver el histograma. El Boxplot a la derecha es la herramienta estándar en ciencia de datos para ver "outliers" (valores extremos o atípicos). Si ves puntos negros a la derecha de la caja, esos son ventas inusualmente altas que podrían distorsionar el promedio.

###Cálculo explícito de métricas:
En lugar de solo usar describe(),se calculó variable por variable (media, mediana, desviación). Esto ayuda a interpretar si la media es representativa o si está sesgada.

###Frecuencia Relativa (%):
En la parte categórica, se agregó el cálculo de porcentajes. Saber que "Electrónica" vendió 150 unidades es útil, pero saber que representa el "30% del total de transacciones" es un dato más valioso para el negocio.

###Ordenamiento de la gráfica:
En el gráfico de barras, se usa el parámetro order para que las barras aparezcan de la más alta a las más bajas, lo cual es una buena práctica de visualización de datos.

##4. Análisis Bivariado (Relación entre dos variables)
Aquí buscamos patrones cruzando datos. Este paso es fundamental porque buscamos relaciones y patrones entre dos variables simultáneamente. Por ejemplo: ¿Cómo cambia la venta según la categoría? ¿Existe relación entre el precio y la cantidad comprada?

Numérico vs Numérico: Precio vs. Cantidad.

Categórico vs Numérico: ¿Qué categoría genera más ingresos? (Boxplot)

```python
# Convertir la columna Fecha a tipo datetime (esencial para analisis temporal)
df['Fecha'] = pd.to_datetime(df['Fecha'])

# Crear una columna de Mes Numerico para ordenar cronologicamente
df['Mes_Num'] = df['Fecha'].dt.month


# ANALISIS BIVARIADO 1: NUMERICO vs CATEGORICO
# Objetivo: Comparar el comportamiento de una variable numerica entre diferentes grupos.
# Ejemplo: Total de Ventas ($) vs Categoria


print("--- 1. ANALISIS: INGRESOS POR CATEGORIA ---")

# Agrupacion de datos para calcular totales exactos
ventas_por_categoria = df.groupby('Categoria')['Total_Venta'].sum().sort_values(ascending=False)
print("Ventas totales acumuladas por Categoria:")
print(ventas_por_categoria)

plt.figure(figsize=(14, 6))

# Grafico A: Boxplot (Diagrama de Caja)
# Nos permite ver la DISPERSION de las ventas dentro de cada categoria.
# ¿Hay categorias que venden montos muy variables o son constantes?
plt.subplot(1, 2, 1)
sns.boxplot(x='Categoria', y='Total_Venta', data=df, palette='Set2', hue='Categoria', legend=False)
plt.title('Distribucion de Ventas por Categoria (Dispersión)')
plt.xlabel('Categoria')
plt.ylabel('Monto por Venta ($)')

# Grafico B: Barplot (Barras de Totales)
# Nos permite ver el VOLUMEN TOTAL de ingresos.
# ¿Que categoria mete mas dinero a la caja al final del dia?
plt.subplot(1, 2, 2)
sns.barplot(x='Categoria', y='Total_Venta', data=df, estimator=sum, palette='viridis', hue='Categoria', legend=False)
plt.title('Ingresos Totales Acumulados por Categoria')
plt.xlabel('Categoria')
plt.ylabel('Total Acumulado ($)')

plt.tight_layout()
plt.show()


# ANALISIS BIVARIADO 2: NUMERICO vs NUMERICO
# Objetivo: Ver si existe correlacion entre dos numeros.
# Ejemplo: Precio Unitario vs Cantidad Comprada


print("\n--- 2. ANALISIS: RELACION PRECIO VS CANTIDAD ---")

plt.figure(figsize=(10, 6))

# Grafico de Dispersion (Scatter Plot)
# Cada punto es una venta.
# Eje X: Precio del producto. Eje Y: Cuantas unidades llevaron.
# Usamos 'hue' para colorear por Categoria y ver si hay grupos diferenciados.
sns.scatterplot(x='Precio_Unitario', y='Cantidad', data=df, hue='Categoria', s=100, alpha=0.7)

plt.title('Correlacion: Precio Unitario vs. Cantidad Comprada')
plt.xlabel('Precio del Producto ($)')
plt.ylabel('Unidades Compradas')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') # Mueve la leyenda afuera

plt.tight_layout()
plt.show()

# Calculo de correlacion matematica (Indice de Pearson)
# 1 = Correlacion perfecta positiva, -1 = Correlacion perfecta negativa, 0 = Sin relacion
correlacion = df[['Precio_Unitario', 'Cantidad', 'Total_Venta']].corr()
print("\nMatriz de Correlacion:")
print(correlacion)



# ANALISIS BIVARIADO 3: SERIE DE TIEMPO (TIEMPO VS NUMERICO)
# Objetivo: Ver la evolucion de una variable a traves del tiempo.
# Ejemplo: Mes vs Ventas Totales


print("\n--- 3. ANALISIS TEMPORAL: TENDENCIA DE VENTAS ---")

# Agrupamos por numero de mes para sumar todo lo vendido en ese periodo
ventas_mensuales = df.groupby('Mes_Num')['Total_Venta'].sum()

plt.figure(figsize=(12, 5))

# Grafico de Linea
# Ideal para ver tendencias (subidas y bajadas) a lo largo del tiempo
plt.plot(ventas_mensuales.index, ventas_mensuales.values, marker='o', linestyle='-', color='darkred', linewidth=2)

plt.title('Evolucion de Ventas Totales (Enero - Noviembre)')
plt.xlabel('Mes del Año (Numero)')
plt.ylabel('Ventas Totales ($)')
plt.grid(True, linestyle='--')
plt.xticks(range(1, 12)) # Asegura que muestre todos los meses del 1 al 11

plt.tight_layout()
plt.show()
```

##Explicaciones adicionales:
##Boxplot vs. Barplot (Análisis Categoría vs. Venta):

Ambos responden preguntas diferentes.

El Barplot responde: "¿Quién vendió más en total?".

El Boxplot responde: "¿Qué tan estables son las ventas?". Si la caja es muy grande, significa que a veces venden muy poco y a veces muchísimo (mucha varianza).

##Scatter Plot (Precio vs. Cantidad):

Este gráfico se usa mucho en economía y retail. Te ayuda a ver la "elasticidad". Normalmente, esperaríamos ver una curva hacia abajo (más precio, menos cantidad),  en datos simulados a veces se ve aleatorio.

hue='Categoria' es para que los puntos tengan colores distintos según si es Ropa, Electrónica, etc., lo que permite encontrar patrones visuales rápidamente.

##Matriz de Correlación (.corr()):

Ej. Si el número entre Precio y Cantidad es cercano a 0, significa que no hay relación lineal clara.

##Serie de Tiempo (Gráfico de Línea):

Es la forma correcta de visualizar datos temporales. marker='o' para resaltar los puntos exactos de cada mes sobre la línea de tendencia.

```python
# =========================================================================
# BLOQUE 1: ANÁLISIS UNIVARIABLE MEJORADO
# Ejecutar este bloque en una celda de Google Colab
# Analiza variables individuales con visualizaciones avanzadas
# =========================================================================

# Importar librerías necesarias (si no están importadas previamente)
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import numpy as np

# Configurar estilo de visualización
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (16, 6)
plt.rcParams['font.size'] = 10

# Calcular estadísticos descriptivos avanzados para Total_Venta
promedio = df['Total_Venta'].mean()
mediana = df['Total_Venta'].median()
desviacion = df['Total_Venta'].std()
minimo = df['Total_Venta'].min()
maximo = df['Total_Venta'].max()
q1 = df['Total_Venta'].quantile(0.25)
q3 = df['Total_Venta'].quantile(0.75)

# Mostrar métricas interpretadas en consola
print("ESTADÍSTICOS DE VENTAS TOTALES")
print(f"Promedio: ${promedio:,.2f}")
print(f"Mediana: ${mediana:,.2f}")
print(f"Desviación Estándar: ${desviacion:,.2f}")
print(f"Rango: ${minimo:,.2f} - ${maximo:,.2f}")
print(f"Cuartil 1 (25%): ${q1:,.2f}")
print(f"Cuartil 3 (75%): ${q3:,.2f}")
print(f"Rango Intercuartílico (IQR): ${q3-q1:,.2f}")

# Crear figura con 3 subplots para análisis completo de variable numérica
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Gráfico 1: Histograma con líneas de referencia estadística
# Muestra distribución de frecuencias con marcadores clave
axes[0].hist(df['Total_Venta'], bins=30, color='#4C72B0', alpha=0.7, edgecolor='black')
axes[0].axvline(promedio, color='red', linestyle='--', linewidth=2, label=f'Media: ${promedio:,.0f}')
axes[0].axvline(mediana, color='green', linestyle='-', linewidth=2, label=f'Mediana: ${mediana:,.0f}')
axes[0].axvline(q1, color='orange', linestyle=':', linewidth=1.5, label=f'Q1: ${q1:,.0f}')
axes[0].axvline(q3, color='purple', linestyle=':', linewidth=1.5, label=f'Q3: ${q3:,.0f}')
axes[0].set_title('Distribución de Ventas Totales\ncon Marcadores Estadísticos')
axes[0].set_xlabel('Monto de Venta ($)')
axes[0].set_ylabel('Frecuencia de Facturas')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Gráfico 2: Boxplot mejorado para detección de outliers
# Identifica valores atípicos que pueden distorsionar el análisis
box_plot = axes[1].boxplot(df['Total_Venta'], patch_artist=True, vert=False, widths=0.6)
box_plot['boxes'][0].set_facecolor('#55A868')
axes[1].set_title('Boxplot de Ventas Totales\n(Detección de Valores Atípicos)')
axes[1].set_xlabel('Monto de Venta ($)')
axes[1].set_yticklabels(['Ventas'])
# Identificar outliers en el gráfico
outliers = box_plot['fliers'][0].get_data()[0]
if len(outliers) > 0:
    for i, outlier in enumerate(outliers[:3]):
        axes[1].text(outlier, 1, f'${outlier:,.0f}', ha='left', va='center', fontsize=8, color='red')

# Gráfico 3: QQ-Plot para validar normalidad de la distribución
# Compara con distribución normal teórica para verificar supuestos estadísticos
stats.probplot(df['Total_Venta'], dist="norm", plot=axes[2])
axes[2].set_title('QQ-Plot: Normalidad de la Distribución')
axes[2].set_xlabel('Cuantiles Teóricos (Normal)')
axes[2].set_ylabel('Cuantiles Observados')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Análisis de variable categórica (Categoria)
print("\nESTADÍSTICOS DE CATEGORÍAS")
conteo_absoluto = df['Categoria'].value_counts()
conteo_relativo = df['Categoria'].value_counts(normalize=True) * 100

# Crear DataFrame de resumen con múltiples métricas
resumen_categoria = pd.DataFrame({
    'Transacciones': conteo_absoluto,
    'Porcentaje': conteo_relativo.round(2),
    'Ingresos_Totales': df.groupby('Categoria')['Total_Venta'].sum(),
    'Ingresos_Promedio_Factura': df.groupby('Categoria')['Total_Venta'].mean()
}).round(2)
resumen_categoria = resumen_categoria.sort_values('Ingresos_Totales', ascending=False)
print(resumen_categoria.to_string())

# Visualización de categorías con doble enfoque
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico 1: Barras horizontales con porcentajes
# Muestra participación relativa de cada categoría
axes[0].barh(resumen_categoria.index, resumen_categoria['Porcentaje'],
             color=['#4C72B0', '#55A868', '#C44E52', '#8172B2'])
for i, pct in enumerate(resumen_categoria['Porcentaje']):
    axes[0].text(pct + 0.5, i, f'{pct}%', va='center', fontsize=11, fontweight='bold')
axes[0].set_title('Participación Porcentual por Categoría')
axes[0].set_xlabel('Porcentaje de Transacciones (%)')
axes[0].grid(True, alpha=0.3, axis='x')

# Gráfico 2: Barras de ingresos totales
# Muestra contribución absoluta al negocio
ingresos_totales = df.groupby('Categoria')['Total_Venta'].sum().sort_values(ascending=False)
bars2 = axes[1].bar(ingresos_totales.index, ingresos_totales.values,
                    color=['#4C72B0', '#55A868', '#C44E52', '#8172B2'])
for i, ingreso in enumerate(ingresos_totales.values):
    axes[1].text(i, ingreso + ingresos_totales.max()*0.01, f'${ingreso:,.0f}',
                 ha='center', va='bottom', fontsize=10, rotation=45)
axes[1].set_title('Ingresos Totales por Categoría')
axes[1].set_ylabel('Ingresos Totales ($)')
axes[1].grid(True, alpha=0.3, axis='y')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# =========================================================================
# BLOQUE 2: ANÁLISIS BIVARIABLE MEJORADO
# Ejecutar este bloque en otra celda de Google Colab
# Analiza relaciones entre dos variables con múltiples perspectivas
# =========================================================================

# Importar librerías necesarias (si no están importadas previamente)
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configurar estilo de visualización
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (20, 10)
plt.rcParams['font.size'] = 10

# Relación Categórica vs Numérica (Categoría vs Total_Venta)
print("ANÁLISIS CATEGORÍA vs TOTAL_VENTA")

# Calcular estadísticos por categoría
stats_categoria = df.groupby('Categoria')['Total_Venta'].agg([
    'mean', 'median', 'std', 'count'
]).round(2)
stats_categoria.columns = ['Promedio', 'Mediana', 'Desviacion', 'Transacciones']
stats_categoria = stats_categoria.sort_values('Promedio', ascending=False)
print("Estadísticos de Ventas por Categoría:")
print(stats_categoria.to_string())

# Crear figura con 3 tipos de visualizaciones comparativas
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Gráfico 1: Boxplot combinado con stripplot
# Muestra distribución completa y puntos individuales
sns.boxplot(x='Categoria', y='Total_Venta', data=df, ax=axes[0],
            palette='Set2', hue='Categoria', legend=False)
sns.stripplot(x='Categoria', y='Total_Venta', data=df, ax=axes[0],
              color='black', alpha=0.4, jitter=0.2, size=3)
axes[0].set_title('Distribución de Ventas por Categoría\n(Boxplot + Puntos)')
axes[0].tick_params(axis='x', rotation=45)
axes[0].set_ylabel('Monto de Venta ($)')

# Gráfico 2: Barras de promedio con intervalos de confianza
# Muestra estimación del valor esperado con incertidumbre
sns.barplot(x='Categoria', y='Total_Venta', data=df, ax=axes[1],
            estimator=np.mean, ci=95, palette='viridis', hue='Categoria', legend=False)
axes[1].set_title('Promedio de Ventas por Categoría\n(Intervalo de confianza 95%)')
axes[1].set_ylabel('Promedio de Venta ($)')
axes[1].tick_params(axis='x', rotation=45)

# Gráfico 3: Violin plot para densidad de probabilidad
# Muestra forma completa de distribución con percentiles
sns.violinplot(x='Categoria', y='Total_Venta', data=df, ax=axes[2],
               palette='pastel', inner='quartile')
axes[2].set_title('Densidad de Ventas por Categoría\n(Violin Plot)')
axes[2].tick_params(axis='x', rotation=45)
axes[2].set_ylabel('Monto de Venta ($)')

plt.tight_layout()
plt.show()

# Relación Numérica vs Numérica (Precio_Unitario vs Cantidad)
print("\nANÁLISIS PRECIO_UNITARIO vs CANTIDAD")

# Calcular coeficiente de correlación
correlacion = df['Precio_Unitario'].corr(df['Cantidad'])
print(f"Coeficiente de correlación: {correlacion:.4f}")

# Matriz de correlación completa
variables_numericas = ['Precio_Unitario', 'Cantidad', 'Total_Venta']
matriz_corr = df[variables_numericas].corr()
print("Matriz de correlación:")
print(matriz_corr.to_string())

# Crear figura con 4 perspectivas de análisis
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Gráfico 1: Scatter plot con regresión lineal
# Muestra tendencia y relación lineal entre variables
sns.regplot(x='Precio_Unitario', y='Cantidad', data=df, ax=axes[0,0],
            scatter_kws={'alpha':0.5, 's':30}, line_kws={'color':'red'})
axes[0,0].set_title('Regresión Lineal: Precio vs Cantidad')
axes[0,0].set_xlabel('Precio Unitario ($)')
axes[0,0].set_ylabel('Cantidad Comprada')

# Gráfico 2: Scatter plot por categoría con colores diferenciados
# Identifica patrones según tipo de producto
sns.scatterplot(x='Precio_Unitario', y='Cantidad', data=df, ax=axes[0,1],
                hue='Categoria', style='Categoria', alpha=0.6, s=60)
axes[0,1].set_title('Precio vs Cantidad por Categoría')
axes[0,1].set_xlabel('Precio Unitario ($)')
axes[0,1].set_ylabel('Cantidad Comprada')
axes[0,1].legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Gráfico 3: Hexbin plot para densidad de puntos
# Útil cuando hay superposición masiva de datos
hb = axes[1,0].hexbin(df['Precio_Unitario'], df['Cantidad'], gridsize=20,
                      cmap='viridis', mincnt=1)
axes[1,0].set_title('Densidad de Puntos: Precio vs Cantidad')
axes[1,0].set_xlabel('Precio Unitario ($)')
axes[1,0].set_ylabel('Cantidad Comprada')
plt.colorbar(hb, ax=axes[1,0], label='Frecuencia')

# Gráfico 4: Heatmap de correlación
# Visualiza fuerza de relación entre todas las variables numéricas
sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', center=0,
            square=True, ax=axes[1,1], fmt='.3f', linewidths=0.5)
axes[1,1].set_title('Matriz de Correlación Completa')

plt.tight_layout()
plt.show()

# Análisis Temporal (Serie de Tiempo)
print("\nANÁLISIS TEMPORAL: EVOLUCIÓN MENSUAL")

# Calcular métricas mensuales agregadas
ventas_mensuales = df.groupby('Mes_Num').agg({
    'Total_Venta': ['sum', 'mean', 'count']
}).round(2)
ventas_mensuales.columns = ['Ingresos_Totales', 'Venta_Promedio', 'Transacciones']

# Crear figura con 3 métricas temporales
fig, axes = plt.subplots(3, 1, figsize=(14, 10))

meses_num = ventas_mensuales.index
ingresos = ventas_mensuales['Ingresos_Totales']
promedios = ventas_mensuales['Venta_Promedio']
transacciones = ventas_mensuales['Transacciones']

# Gráfico 1: Ingresos totales mensuales con relleno
axes[0].plot(meses_num, ingresos, marker='o', linestyle='-', linewidth=2,
             markersize=8, color='#4C72B0')
axes[0].fill_between(meses_num, ingresos, alpha=0.3, color='#4C72B0')
for x, y in zip(meses_num, ingresos):
    axes[0].text(x, y, f'${y:,.0f}', ha='center', va='bottom', fontsize=9)
axes[0].set_title('Ingresos Totales Mensuales (Ene-Nov 2025)')
axes[0].set_ylabel('Ingresos Totales ($)')
axes[0].grid(True, alpha=0.3)

# Gráfico 2: Número de transacciones por mes
axes[1].bar(meses_num, transacciones, color='#55A868', alpha=0.8)
for x, y in zip(meses_num, transacciones):
    axes[1].text(x, y, f'{int(y)}', ha='center', va='bottom', fontsize=9)
axes[1].set_title('Cantidad de Transacciones por Mes')
axes[1].set_ylabel('Número de Facturas')
axes[1].set_xlabel('Mes del Año')
axes[1].grid(True, alpha=0.3, axis='y')

# Gráfico 3: Venta promedio por factura
axes[2].plot(meses_num, promedios, marker='s', linestyle='--', linewidth=2,
             markersize=7, color='#C44E52')
axes[2].set_title('Monto Promedio por Factura')
axes[2].set_ylabel('Venta Promedio ($)')
axes[2].set_xlabel('Mes del Año')
axes[2].grid(True, alpha=0.3)

# Etiquetas de meses para todos los gráficos
meses_labels = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun',
                'Jul', 'Ago', 'Sep', 'Oct', 'Nov']
for ax in axes:
    ax.set_xticks(meses_num)
    ax.set_xticklabels(meses_labels)

plt.tight_layout()
plt.show()

# Análisis de Top 10 Clientes
print("\nANÁLISIS TOP 10 CLIENTES")

# Calcular métricas por cliente
top_clientes = df.groupby('Cliente')['Total_Venta'].agg([
    'sum', 'mean', 'count'
]).round(2)
top_clientes.columns = ['Ingreso_Total', 'Venta_Promedio', 'Frecuencia']
top_clientes = top_clientes.sort_values('Ingreso_Total', ascending=False).head(10)
print("Top 10 Clientes:")
print(top_clientes.to_string())

# Visualización de clientes
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Gráfico 1: Ingresos totales del top 10
clientes = top_clientes.index
ingresos_clientes = top_clientes['Ingreso_Total']
axes[0].barh(clientes, ingresos_clientes, color='#4C72B0')
for i, ingreso in enumerate(ingresos_clientes):
    axes[0].text(ingreso, i, f'${ingreso:,.0f}', va='center', ha='left', fontsize=9)
axes[0].set_title('Top 10 Clientes - Ingresos Totales')
axes[0].set_xlabel('Ingresos Totales ($)')

# Gráfico 2: Frecuencia vs. Ingreso promedio (burbujas)
# Tamaño de burbuja representa ingreso total
scatter = axes[1].scatter(top_clientes['Frecuencia'], top_clientes['Venta_Promedio'],
                          s=top_clientes['Ingreso_Total']/1000, alpha=0.6,
                          c=range(len(top_clientes)), cmap='viridis')
axes[1].set_title('Frecuencia vs. Venta Promedia\n(Tamaño de burbuja = Ingreso Total)')
axes[1].set_xlabel('Número de Compras')
axes[1].set_ylabel('Venta Promedio por Factura ($)')
# Etiquetas de clientes
for i, cliente in enumerate(clientes):
    axes[1].annotate(cliente.split()[0],
                     (top_clientes.iloc[i]['Frecuencia'], top_clientes.iloc[i]['Venta_Promedio']),
                     fontsize=8, alpha=0.8)
plt.colorbar(scatter, ax=axes[1], label='Ranking')

plt.tight_layout()
plt.show()
```
